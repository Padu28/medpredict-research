# 02 · Baseline and model comparison

```python
import json

import matplotlib.pyplot as plt
import pandas as pd

from medpredict.common.config import get_settings
from medpredict.storage.database import Database

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 20)
db = Database(get_settings().db_path)
r = db.query_df("SELECT model, calibration, metrics_json FROM model_results")
tbl = pd.DataFrame([{"model": a, "calibration": b, **{k: v for k, v in json.loads(j).items() if not isinstance(v, (list, dict))}}
                    for a, b, j in zip(r.model, r.calibration, r.metrics_json)])
tbl[tbl.calibration == "uncalibrated"][["model", "auroc", "pr_auc", "brier", "ece", "accuracy_at_0.5"]].round(4)
```

AUROC differences between the interpretable baseline and the tree ensembles are within the bootstrap
confidence intervals — the simpler model is competitive.
